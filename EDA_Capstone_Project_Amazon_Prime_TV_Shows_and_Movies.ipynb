# **Remove Specifice symbols**

```python
def clean_text(text):
  if pd.isna(text):
    return ""
  text=text.lower()
  text=text.strip()
  text=text.replace('!', '').replace('#', '')
  for char in ['@','$','%','^','&','*','(', ')','-','_','=','+','[', ']','{', '}','/','|','?',';',':','"','<','.','>','\\','/']:
    text =text.replace(char, '')

  return ''.join([i for i in text if not i.isdigit()])
```

```python
data['genres']=data['genres'].apply(clean_text)
```

```python
data['description'].unique()
```

```python
data['description']=data['description'].apply(clean_text)
```

# **Check the Outlier of Box plot**
**Column name are (runtime)**

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=data, x='runtime', color='red')
plt.tight_layout()
plt.show()
```

# **Outliear Treatment are IQR Method :-**



1.   Capping
2.   Flooring





```python
Q1=data['runtime'].quantile(0.25) # 25% of percentage values
Q3=data['runtime'].quantile(0.75) # 75% of percentage values
IQR=Q3-Q1                         # IQR =  75% of percentage values -  25% of percentage values
print(IQR)
```

```python
upper_bond=Q3 + 1.5*IQR
lower_bond=Q1 - 1.5*IQR
print(upper_bond)
print(lower_bond)
```

```python
new_data=data.copy()
new_data['runtime']=np.where(
    new_data['runtime'] > upper_bond,
    upper_bond,
    np.where(
        new_data['runtime'] < lower_bond,
        lower_bond,
        new_data['runtime']
    )
)
```

# **Clean the Outlier**

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=new_data, x='runtime', color='red')
plt.tight_layout()
plt.show()
```
